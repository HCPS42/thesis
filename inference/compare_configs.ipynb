```python
import os

run_names = [dirname for dirname in os.listdir('logs/') if os.path.isdir(os.path.join('logs/', dirname))]
print(run_names)
```

```python
from graphs import process_run
import pandas as pd

data = []

for run_name in run_names:
    df, summary, lengths, times = process_run(run_name, n_bootstrap=None)
    
    aime_2024_pass1_accuracy = df[df['source'] == 'AIME 2024']['accuracy'].mean()
    aime_2025_pass1_accuracy = df[df['source'] == 'AIME 2025']['accuracy'].mean()
    
    aime_2024_majK_accuracy = (df[df['source'] == 'AIME 2024']['solved']).mean() * 30
    aime_2025_majK_accuracy = (df[df['source'] == 'AIME 2025']['solved']).mean() * 30
    
    data.append({
        'run_name': run_name, 
        'AIME 2024 (pass@1)': aime_2024_pass1_accuracy, 
        'AIME 2025 (pass@1)': aime_2025_pass1_accuracy,
        'AIME 2024 (maj@K)': aime_2024_majK_accuracy,
        'AIME 2025 (maj@K)': aime_2025_majK_accuracy
    })

result_df = pd.DataFrame(data)
result_df
```
